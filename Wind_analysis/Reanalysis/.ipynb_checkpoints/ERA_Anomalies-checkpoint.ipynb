# Imports

```python
import xarray as xr
from dask_jobqueue import PBSCluster
from dask.distributed import Client
```

# PBS

```python
cluster = PBSCluster(
    cores=1, # The number of cores you want
    memory='64GB', # Amount of memory (resource_spec is the one)
    processes=1, # How many processes
    queue='casper', # The type of queue to utilize (/glade/u/apps/dav/opt/usr/bin/execcasper)
    local_directory='$TMPDIR', # Use your local directory
    resource_spec='select=1:ncpus=1:mem=64GB', # Specify resources
    account='P93300313', # Input your project ID here
    walltime='01:00:00', # Amount of wall time
    n_workers=3,
)
# Setup your client
client = Client(cluster)
```

```python
client#.shutdown()
```

# Load

```python
ds = xr.open_dataset('/glade/work/acruz/Reanalysis/ERA5_circulation.nc', chunks={'time': 528, 'level': 37, 'latitude': 81, 'longitude': 801})
ds
```

# Calculate

```python
clim = ds.groupby('time.month').mean(dim='time')
clim.compute()
clim
```

```python
circ_anom = ds.groupby('time.month') - clim
circ_anom.compute()
circ_anom
```

```python
# meridional mean
mean_anom = circ_anom.mean(dim='latitude').compute()
mean_anom
```

```python
mean_anom.to_netcdf('/glade/work/acruz/Reanalysis/ERA5_UW_tropic_anom.nc')
```
