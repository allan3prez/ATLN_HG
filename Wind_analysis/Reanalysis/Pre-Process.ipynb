# Circulation data
Join all the files in the reanalysis, reduce to scope (10S to 10N, 180W to 20E), output single ds with U, W

```python
import xarray as xr
import numpy as np
import glob
```

```python
ufpath = '/gdex/data/d633001/e5.moda.an.pl/*/e5.moda.an.pl.128_131_u*.nc'
ufiles = glob.glob(ufpath)
uds = xr.open_mfdataset(ufiles)
uds
```

```python
wfpath = '/gdex/data/d633001/e5.moda.an.pl/*/e5.moda.an.pl.128_135_w*.nc'
wfiles = glob.glob(wfpath)
wds = xr.open_mfdataset(wfiles)
wds
```

```python
wds
```

```python
ds = uds.merge(wds)
ds
```

```python
ds.to_netcdf('/glade/work/acruz/Reanalysis/ERA5_UW.nc')
```
