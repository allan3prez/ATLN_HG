# Circulation data
Join all the files in the reanalysis, reduce to scope (10S to 10N, 180W to 20E), output single ds with U, W

```python
import xarray as xr
import numpy as np
import glob
from dask_jobqueue import PBSCluster
from dask.distributed import Client
```

```python
cluster = PBSCluster(
    cores=1, # The number of cores you want
    memory='64GB', # Amount of memory (resource_spec is the one)
    processes=1, # How many processes
    queue='casper', # The type of queue to utilize (/glade/u/apps/dav/opt/usr/bin/execcasper)
    local_directory='$TMPDIR', # Use your local directory
    resource_spec='select=1:ncpus=1:mem=64GB', # Specify resources
    account='P93300313', # Input your project ID here
    walltime='01:00:00', # Amount of wall time
    n_workers=7,
)
# Setup your client
client = Client(cluster)
```

```python
client
```

```python
ufpath = '/gdex/data/d633001/e5.moda.an.pl/*/e5.moda.an.pl.128_131_u*.nc'
ufiles = glob.glob(ufpath)
uds = xr.open_mfdataset(ufiles)
uds = uds.chunk({'time': 96, 'latitude': 721, 'longitude': 1440})
uds
```

```python
wfpath = '/gdex/data/d633001/e5.moda.an.pl/*/e5.moda.an.pl.128_135_w*.nc'
wfiles = glob.glob(wfpath)
wds = xr.open_mfdataset(wfiles)
wds = wds.chunk({'time': 96, 'latitude': 721, 'longitude': 1440})
wds
```

```python
ds = uds.merge(wds)
ds
```

```python
ds = ds.chunk({'time': 96, 'latitude': 721, 'longitude': 1440})
ds
```

```python
ds.coords['longitude'] = (ds.coords['longitude'] + 180) % 360 - 180
ds = ds.sortby(ds.longitude)
```

```python
# making sure the latitude transformation works properly
newlats = (ds.coords['latitude'] + 90) % 180 - 90
newlats.sortby(newlats.latitude)
```

```python
ds = ds.sortby(newlats.latitude)
ds
```

```python
tropic = ds.sel(latitude=slice(-10, 10), longitude=slice(-180, 20))
tropic.compute()
tropic
```

```python
tropic.to_netcdf('/glade/work/acruz/Reanalysis/ERA5_circulation.nc')
```
